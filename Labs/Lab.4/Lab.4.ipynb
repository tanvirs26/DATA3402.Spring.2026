7. Add a function to the object classes, including the base, that returns a list of up to 16 pairs of  $x$ and $y$ points on the parameter of the object. 

```python
class Shape:
    def get_x(self):
        raise NotImplementedError
    
    def get_y(self):
        raise NotImplementedError
    
    def area(self):
        raise NotImplementedError
    
    def perimeter(self):
        raise NotImplementedError
    
    def boundary_points(self, n=16):
        raise NotImplementedError
    
class Rectangle(Shape):
    def __init__(self, l, w, x, y):
        self.__l = l
        self.__w = w
        self.__x = x
        self.__y = y
    
    def get_length(self):
        return self.__l
    
    def get_width(self):
        return self.__w
    
    def get_x(self):
        return self.__x
    
    def get_y(self):
        return self.__y
    
    def area(self):
        return self.__l * self.__w
    
    def perimeter(self):
        return (2 * self.__l) + (2 * self.__w)
    
    def boundary_points(self, n=16):
        n = min(n, 16)
        points = []

        x0, y0 = self.__x, self.__y
        l, w = self.__l, self.__w

        per = self.perimeter()
        step = per / n

        for i in range(n):
            d = i * step

            if d < l:
                points.append((x0 + d, y0))
            elif d < l + w:
                points.append((x0 + l, y0 + (d - l)))
            elif d < 2*l + w:
                points.append((x0 + l - (d - (l + w)), y0 + w))
            else:
                points.append((x0, y0 + w - (d - (2*l + w))))

        return points

class Circle(Shape):
    def __init__(self, r, x, y):
        self.__r = r
        self.__x = x
        self.__y = y

    def get_radius(self):
        return self.__r
    
    def get_x(self):
        return self.__x
    
    def get_y(self):
        return self.__y
    
    def area(self):
        return math.pi * (self.__r ** 2)
    
    def perimeter(self):
        return 2 * math.pi * self.__r
    
    def boundary_points(self, n=16):
        n = min(n, 16)
        points = []

        for i in range(n):
            theta = 2 * math.pi * i / n
            x = self.__x + self.__r * math.cos(theta)
            y = self.__y + self.__r * math.sin(theta)
            points.append((x, y))

        return points
    
class Triangle(Shape):
    def __init__(self, a, base, c, h, x, y):
        self.__a = a
        self.__base = base
        self.__c = c
        self.__h = h
        self.__x = x
        self.__y = y

    def get_side_a(self):
        return self.__a
    
    def get_base(self):
        return self.__base
    
    def get_side_c(self):
        return self.__c
    
    def get_height(self):
        return self.__h
    
    def get_x(self):
        return self.__x
    
    def get_y(self):
        return self.__y
    
    def area(self):
        return (1/2) * self.__base * self.__h
    
    def perimeter(self):
        return self.__a + self.__base + self.__c
    
    def boundary_points(self, n=16):
        n = min(n, 16)
        points = []

        A = (self.__x, self.__y)
        B = (self.__x + self.__base, self.__y)
        C = (self.__x, self.__y + self.__h)

        edges = [(A, B), (B, C), (C, A)]
        per = self.perimeter()
        step = per / n

        lengths = []
        for p1, p2 in edges:
            lengths.append(math.dist(p1, p2))

        for i in range(n):
            d = i * step
            acc = 0

            for (p1, p2), length in zip(edges, lengths):
                if acc + length >= d:
                    t = (d - acc) / length
                    x = p1[0] + t * (p2[0] - p1[0])
                    y = p1[1] + t * (p2[1] - p1[1])
                    points.append((x, y))
                    break
                acc += length

        return points
```

```python
class Shape:
    def contains_point(self, x, y):
        raise NotImplementedError

class Rectangle(Shape):
    def __init__(self, l, w, x, y):
        self.__l = l
        self.__w = w
        self.__x = x
        self.__y = y
    
    def get_length(self):
        return self.__l
    
    def get_width(self):
        return self.__w
    
    def get_x(self):
        return self.__x
    
    def get_y(self):
        return self.__y
    
    def area(self):
        return self.__l * self.__w
    
    def perimeter(self):
        return (2 * self.__l) + (2 * self.__w)
    
    def contains_point(self, x, y):
        return (self.get_x() <= x <= self.get_x() + self.get_length() and
                self.get_y() <= y <= self.get_y() + self.get_width())

class Circle(Shape):
    def __init__(self, r, x, y):
        self.__r = r
        self.__x = x
        self.__y = y

    def get_radius(self):
        return self.__r
    
    def get_x(self):
        return self.__x
    
    def get_y(self):
        return self.__y
    
    def area(self):
        return math.pi * (self.__r ** 2)
    
    def perimeter(self):
        return 2 * math.pi * self.__r
    
    def contains_point(self, x, y):
        dx = x - self.get_x()
        dy = y - self.get_y()
        return dx**2 + dy**2 <= self.get_radius()**2
    
class Triangle(Shape):
    def __init__(self, a, base, c, h, x, y):
        self.__a = a
        self.__base = base
        self.__c = c
        self.__h = h
        self.__x = x
        self.__y = y

    def get_side_a(self):
        return self.__a
    
    def get_base(self):
        return self.__base
    
    def get_side_c(self):
        return self.__c
    
    def get_height(self):
        return self.__h
    
    def get_x(self):
        return self.__x
    
    def get_y(self):
        return self.__y
    
    def area(self):
        return (1/2) * self.__base * self.__h
    
    def perimeter(self):
        return self.__a + self.__base + self.__c
    
    def contains_point(self, x, y):
        x0, y0 = self.get_x(), self.get_y()
        x1, y1 = x0 + self.get_base(), y0
        x2, y2 = x0, y0 + self.get_height()
        
        denom = (y1 - y2)*(x0 - x2) + (x2 - x1)*(y0 - y2)
        a = ((y1 - y2)*(x - x2) + (x2 - x1)*(y - y2)) / denom
        b = ((y2 - y0)*(x - x2) + (x0 - x2)*(y - y2)) / denom
        c = 1 - a - b
        return 0 <= a <= 1 and 0 <= b <= 1 and 0 <= c <= 1
```

```python
class Shape:
    def contains_point(self, x, y):
        raise NotImplementedError
    
    def boundary_points(self, n=16):
        raise NotImplementedError
    
    def paint(self, canvas, char='*'):
        raise NotImplementedError
    
class Canvas:
    def __init__(self, width, height):
        self.width = width
        self.height = height
        # Empty canvas is a matrix with element being the "space" character
        self.data = [[' '] * width for i in range(height)]

    def set_pixel(self, row, col, char='*'):
        self.data[row][col] = char

    def get_pixel(self, row, col):
        return self.data[row][col]
    
    def clear_canvas(self):
        self.data = [[' '] * self.width for i in range(self.height)]
    
    def v_line(self, x, y, w, **kargs):
        for i in range(x,x+w):
            self.set_pixel(i,y, **kargs)

    def h_line(self, x, y, h, **kargs):
        for i in range(y,y+h):
            self.set_pixel(x,i, **kargs)
            
    def line(self, x1, y1, x2, y2, **kargs):
        slope = (y2-y1) / (x2-x1)
        for y in range(y1,y2):
            x= int(slope * y)
            self.set_pixel(x,y, **kargs)
            
    def display(self):
        print("\n".join(["".join(row) for row in self.data]))

class Rectangle(Shape):
    def __init__(self, l, w, x, y, name=""):
        self.__l = l
        self.__w = w
        self.__x = x
        self.__y = y
        self.name = name
    
    def contains_point(self, x, y):
        return (self.__x <= x <= self.__x + self.__l and
                self.__y <= y <= self.__y + self.__w)
    
    def boundary_points(self, n=16):
        points = []
        l, w, x0, y0 = self.__l, self.__w, self.__x, self.__y
        perim = 2*(l + w)
        step = perim / n
        for i in range(n):
            d = i * step
            if d < l:
                points.append((x0 + d, y0))
            elif d < l + w:
                points.append((x0 + l, y0 + (d - l)))
            elif d < 2*l + w:
                points.append((x0 + l - (d - (l + w)), y0 + w))
            else:
                points.append((x0, y0 + w - (d - (2*l + w))))
        return points
    
    def paint(self, canvas, char='*'):
        x0, y0 = int(self.__x), int(self.__y)
        for i in range(int(self.__l)):
            canvas.set_pixel(y0, x0 + i, char)
            canvas.set_pixel(y0 + int(self.__w) - 1, x0 + i, char)
        for j in range(int(self.__w)):
            canvas.set_pixel(y0 + j, x0, char)
            canvas.set_pixel(y0 + j, x0 + int(self.__l) - 1, char)

class Circle(Shape):
    def __init__(self, r, x, y, name=""):
        self.__r = r
        self.__x = x
        self.__y = y
        self.name = name
    
    def contains_point(self, x, y):
        dx = x - self.__x
        dy = y - self.__y
        return dx**2 + dy**2 <= self.__r**2
    
    def boundary_points(self, n=16):
        points = []
        for i in range(n):
            theta = 2 * math.pi * i / n
            points.append((self.__x + self.__r * math.cos(theta),
                           self.__y + self.__r * math.sin(theta)))
        return points
    
    def paint(self, canvas, char='*'):
        for y in range(canvas.height):
            for x in range(canvas.width):
                if self.contains_point(x, y):
                    canvas.set_pixel(y, x, char)

class Triangle(Shape):
    def __init__(self, a, base, c, h, x, y, name=""):
        self.__a = a
        self.__base = base
        self.__c = c
        self.__h = h
        self.__x = x
        self.__y = y
        self.name = name
    
    def contains_point(self, x, y):
        x0, y0 = self.__x, self.__y
        x1, y1 = x0 + self.__base, y0
        x2, y2 = x0, y0 + self.__h
        denom = (y1 - y2)*(x0 - x2) + (x2 - x1)*(y0 - y2)
        a = ((y1 - y2)*(x - x2) + (x2 - x1)*(y - y2)) / denom
        b = ((y2 - y0)*(x - x2) + (x0 - x2)*(y - y2)) / denom
        c = 1 - a - b
        return 0 <= a <= 1 and 0 <= b <= 1 and 0 <= c <= 1
    
    def boundary_points(self, n=16):
        x0, y0 = self.__x, self.__y
        x1, y1 = x0 + self.__base, y0
        x2, y2 = x0, y0 + self.__h
        edges = [(x0,y0,x1,y1), (x1,y1,x2,y2), (x2,y2,x0,y0)]
        points = []
        for i in range(n):
            edge = edges[i % 3]
            t = (i // 3) / (n/3)
            x = edge[0] + t*(edge[2]-edge[0])
            y = edge[1] + t*(edge[3]-edge[1])
            points.append((x,y))
        return points
    
    def paint(self, canvas, char='*'):
        for y in range(canvas.height):
            for x in range(canvas.width):
                if self.contains_point(x, y):
                    canvas.set_pixel(y, x, char)

class CompoundShape(Shape):
    def __init__(self, shapes=None, name=""):
        self.shapes = shapes if shapes is not None else []
        self.name = name
    
    def add_shape(self, shape):
        self.shapes.append(shape)
    
    def contains_point(self, x, y):
        return any(s.contains_point(x, y) for s in self.shapes)
    
    def boundary_points(self, n=16):
        points = []
        for s in self.shapes:
            points.extend(s.boundary_points(n))
        return points[:16] 
    
    def paint(self, canvas, char='*'):
        for s in self.shapes:
            s.paint(canvas, char)
```
